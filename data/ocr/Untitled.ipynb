```python
import pytesseract
import pandas as pd
import cv2
import numpy as np
import re
from PIL import Image

# Load the image
image_path = "./images/43.png"
image = Image.open(image_path)

# Convert image to OpenCV format
image_cv = np.array(image)

# Convert to grayscale
gray = cv2.cvtColor(image_cv, cv2.COLOR_RGB2GRAY)

# Apply noise reduction (Gaussian Blur)
gray = cv2.GaussianBlur(gray, (3, 3), 0)

# Apply adaptive thresholding to enhance text visibility
thresh = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

# Apply morphological transformations to remove small noise
kernel = np.ones((1, 1), np.uint8)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Perform OCR using Vietnamese language model
custom_config = "--psm 6 --oem 3 -l vie"
extracted_text = pytesseract.image_to_string(thresh, config=custom_config)

# Clean extracted text (remove unnecessary spaces & multiple newlines)
extracted_text = re.sub(r'\n+', '\n', extracted_text).strip()

# Print raw extracted text for debugging
print("Raw OCR Output:\n", extracted_text)

# Split text into lines and extract structured data
lines = extracted_text.split("\n")
structured_data = []

# Define regex pattern for structured row detection
row_pattern = re.compile(r"(\d{10,})\s+([\w\sÀ-Ỹà-ỹ.,'-]+)\s+(\d{2}/\d{2}/\d{4})\s+([\w\sÀ-Ỹà-ỹ.,'-]+)\s+(\d{4,}-\d{5})\s+([\w\sÀ-Ỹà-ỹ.,'-]+)")

for line in lines:
    match = row_pattern.search(line)
    if match:
        structured_data.append(match.groups())

# Convert structured data into a Pandas DataFrame
df = pd.DataFrame(structured_data, columns=["ID", "Name", "DOB", "Parent Name", "Code", "Other Info"])

# Display the structured DataFrame
import ace_tools as tools  # For displaying tables
tools.display_dataframe_to_user(name="Optimized OCR Extracted Table", dataframe=df)
```
