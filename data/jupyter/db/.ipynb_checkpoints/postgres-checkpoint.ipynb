```python
from sqlalchemy import create_engine
import pandas as pd

# Create a connection string
conn_string = "postgresql+psycopg2://postgres:postgres@localhost:5432/tthq"

# Create an engine
engine = create_engine(conn_string)

# Query the database and load into a DataFrame
query = "SELECT * FROM ttqh_phuong_xa"
df = pd.read_sql(query, engine)

# Display the DataFrame
print(df.head())
```

```python
# Basic statistics
print(df.describe())

# Filter data
filtered_data = df[df['id'] > 100]

# Group by and aggregate
grouped_data = df.groupby('maquanhuyen')

# Visualize data
import matplotlib.pyplot as plt
grouped_data['maquanhuyen'].hist(bins=20)
plt.show()
```
